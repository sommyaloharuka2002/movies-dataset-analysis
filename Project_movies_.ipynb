```python
# Check if any NANs in your dataset and fill them with a good filer
print(df.isnull().sum(),"\n")    # Check for NANs in each column
df.fillna({'homepage': 'Unknown','overview': 'No overview', 'release_date': '---', 'runtime': 'XX:XX:XX','tagline': 'No tagline' }, inplace=True)
df
```

```python
#In the given movie dataset, identify all the rows in the runtime column that contain the keyword 'XX:XX:XX', regardless of the case. Provide all the rows that match this condition
df[df['runtime'].str.contains('XX:XX:XX', case=False, na=False)]   #case=False makes it case insensitive, which means it will match 'XX:XX:XX', 'xx:xx:xx', 'Xx:Xx:Xx', etc. and na=False ensures that any NaN values in the 'runtime' column are treated as False and do not cause errors during the search.
```

```python
#drop homepage (unnecessary column)
df.drop('homepage', axis=1, inplace=True)  #axis=1 indicates that we are dropping a column (as opposed to axis=0, which would indicate dropping a row)
                             #inplace=True can be added to make the change permanent in the dataframe
```

```python
#impute null values in tagline column with unknown
df['tagline'] = df['tagline'].replace('No tagline', 'Unknown')     #to replace 'No tagline' with 'Unknown'
count = np.sum(df['tagline'] == 'Unknown')
count
```

```python
# Drop all the null values in the dataset
#df.dropna()   # since we have already filled the null values, this will not drop any rows
df.isnull().sum()
```

```python
# Discover the types of each columns and modify them if needed
df['release_date'] = pd.to_datetime(df['release_date'], errors='coerce')  # to check if the conversion is successful)
df['status'] = df['status'].astype('category')  # to change the datatype to category [astype() function is used to cast a pandas object to a specified data type.]
df['original_language'] = df['original_language'].astype('category')  # to change the datatype to category
df['runtime'] = pd.to_numeric(df['runtime'], errors='coerce')  # to convert runtime to numeric(float/int) format
df
```

```python
# Check the conversion in the particular columns and their rows with null values(if any)
df.loc[df['runtime'].isnull()]  # to check if there are any null values after conversion
```

```python
# Give a very simple statistical analsys for the numerical columns
df.describe().T   #.T is used to transpose the dataframe for better readability
df
```

```python
#create a release_year from the date values in release_date column

df['release_year'] = df['release_date'].dt.year # Extract year
df[['release_year', 'release_date']]
df.dtypes
```

```python
# Calculate the mean rate for [1999, 1980, 2004, 2017] the movies grouped by years
years = [1999, 1980, 2004, 2017]

result = df[df['release_year'].isin(years)].groupby('release_year', as_index=False)['vote_average'].mean()
print(result)

# Calculate the average rate for movies released in 2009
print("\n \nAverage rate for movies released in 2009 is:", df[df['release_year'] == 2009]['vote_average'].mean().round(6))
```

```python
# Rearange the dataframe based on revenue - budget values
df['Profit'] = df['revenue'] - df['budget']
df_sorted = df.sort_values(by=['Profit'], ascending=False)
df_sorted.head()
```

```python
# Reformat the movie genres values

import ast
df.loc[:, 'genres'] = df['genres'].apply(lambda x: [i['name'] for i in ast.literal_eval(x)])
df[['title', 'genres']]
```
